```python
# Function to clean raw transaction data for RFM segmentation
def clean_data(df):
    # Step 1: Remove transactions with missing CustomerID (cannot segment anonymous users)
    df = df.dropna(subset = ["CustomerID"])
    
    # Step 2: Keep only valid purchases (positive quantity, non-zero price, exclude cancellations)
    df= df.query("Quantity>0 and UnitPrice>0 and not InvoiceNo.str.startswith('C')")
    
    #Step 3 : Filtering non-products
    non_products = ['DOT', 'M', 'C2', 'POST', 'BANK CHARGES', 'PADS', 'CRUK']
    df = df[~df['StockCode'].isin(non_products)]
    
    return df
    
# Load original dataset
df = pd.read_csv("online_retail.csv")

# Execute data cleaning pipeline
cleaned_data = clean_data(df)    
```

```python
cleaned_data["TotalSpend"] = cleaned_data["Quantity"]*cleaned_data["UnitPrice"]
```

```python
cleaned_data["InvoiceDate"] =pd.to_datetime(cleaned_data["InvoiceDate"])
```

```python
cleaned_data.info()
```

```python
reference_date = cleaned_data["InvoiceDate"].max() + pd.Timedelta(days=1)
```

```python
rfm = cleaned_data.groupby('CustomerID').agg({
    "InvoiceDate" : lambda x:(reference_date - x.max()).days,
    "InvoiceNo" : "unique",
    "TotalSpend" : "sum"
        }).reset_index()
    
```

```python
rfm
```

```python
rfm["R_Score"] = pd.qcut(rfm["InvoiceDate"],q = 5, labels = [5,4,3,2,1])
```

```python
rfm["R_Score"]
```

```python
rfm["InvoiceNo"] = rfm["InvoiceNo"].str.len()
```

```python
rfm["F_Score"] = pd.qcut(rfm["InvoiceNo"].rank(method='first').values,q = 5 , labels = [1,2,3,4,5])
```

```python
rfm
```

```python
rfm["M_Score"] = pd.qcut(rfm["TotalSpend"],q = 5 , labels = [1,2,3,4,5])
```

```python
rfm
```

```python
rfm["RFM_Score"] = (rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str))
```

```python
rfm
```

```python
rfm.info()
```

```python
def assign_segment(row):
    r = int(row['R_Score'])
    f = int(row['F_Score'])
    
    if r >= 4 and f >= 4:
        return 'Champions'
    elif r >= 3 and (f == 2 or f == 3):
        return 'Potential Loyalists'
    elif (r == 4 or r == 5) and f == 1:
        return 'New Customers'
    elif (r == 1 or r == 2) and f >= 4:
        return "At Risk (Can't Lose Them)"
    elif r == 1 and f == 1:
        return 'Lost'
    else:
        return 'Others'

# Appling row-wise
rfm['Segment'] = rfm.apply(assign_segment, axis=1)
```

```python
rfm
```

```python
Segment_Profile = rfm.groupby("Segment").agg({
    "InvoiceDate" : "mean",
    "InvoiceNo" : "mean",
    "TotalSpend" : ["mean","sum"],
    "CustomerID" : "count"
})
```
